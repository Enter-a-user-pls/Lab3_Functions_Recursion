```python
# [RECURSIVE FAULT TRACE]

from Student_Input import *

# ==========================================================
system_id = int(BYTEVAL)        # which system/module the fault belongs to
severity = int(SEED_NUM)        # how serious the fault is
error_depth = int(ARTIST_NUMBER)      # how many layers deep the fault propagates

# ==========================================================
# TRACKERS
call_counter = 0     # counts every recursive call
execution_log = []   # stores the step-by-step log


def generate_fault_code(system_id, severity, error_depth):
    """
    Combine the 3 variables into one unique fault code.
    Format: F<system_id>-<severity>-<error_depth>-<checksum>
    The checksum makes the code unique to this exact combination.
    """
    checksum = (system_id * 100) + (severity * 10) + error_depth
    return f"F{system_id}-{severity}-{error_depth}-{checksum}"


def trace_fault(fault_code, layer, severity):
    """
    Recursive function that traces the fault layer by layer.

    Base condition: layer == 0 (the fault has reached the root cause).
    Recursive case: move one layer deeper and reduce the remaining
                    severity impact by 1 (minimum 0).
    """
    global call_counter
    call_counter += 1

    # ---- BASE CONDITION ----
    if layer == 0:
        execution_log.append(
            f"Call {call_counter}: layer={layer} -> BASE CONDITION reached "
            f"(root cause identified)"
        )
        return [f"Layer 0: ROOT CAUSE of {fault_code}"]

    # ---- RECURSIVE CASE ----
    remaining_severity = max(severity - 1, 0)
    execution_log.append(
        f"Call {call_counter}: layer={layer}, severity={severity} "
        f"-> descending to layer {layer - 1}"
    )

    current_step = (
        f"Layer {layer}: fault {fault_code} propagating "
        f"(severity impact = {severity})"
    )

    # Recursive call on the next (lower) layer
    return [current_step] + trace_fault(fault_code, layer - 1, remaining_severity)


def main():
    # 1. Generate the fault code
    fault_code = generate_fault_code(system_id, severity, error_depth)

    # 2. Run the recursive trace
    trace = trace_fault(fault_code, error_depth, severity)

    # 3. Display everything in the required format
    print("generated fault data:")
    print(f"  system_id   = {system_id}")
    print(f"  severity    = {severity}")
    print(f"  error_depth = {error_depth}")
    print(f"  fault code  = {fault_code}")
    print()

    print("recursive trace:")
    for step in trace:
        print(f"  {step}")
    print()

    print("number of recursive calls:")
    print(f"  {call_counter}")
    print()

    print("execution log:")
    for entry in execution_log:
        print(f"  {entry}")
    print()

    print("final output:")
    print(
        f"  Fault {fault_code} traced through {error_depth} layers to its "
        f"root cause using {call_counter} recursive calls."
    )


if __name__ == "__main__":
    main()

```
